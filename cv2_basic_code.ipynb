```python
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix
from sklearn.model_selection import train_test_split as tts
from sklearn.neighbors import KNeighborsClassifier as KNN
all_images=[]
all_labels=[]
base_folder_path=r'C:\Users\akatb\OneDrive\Documents\ASL\American'
chars="abcdefghijklmnopqrstuvwxyz0123456789"
label_mapping={}
index=0
for char in chars:
    label_mapping[char]=index
    index += 1
print("Loading and preprocessing images...")
print(label_mapping)
for folder in label_mapping:
    folder_path = os.path.join(base_folder_path, folder)
    print(folder_path)
    if not os.path.exists(folder_path): continue

    for image_name in os.listdir(folder_path):
        try:
            img = cv2.imread(os.path.join(folder_path, image_name), cv2.IMREAD_GRAYSCALE)
            img = cv2.resize(img, (64, 64))

            all_images.append(img.flatten())
            all_labels.append(label_mapping[folder])
        except Exception:
            continue
print('And now for the easy part!')
all_images=np.array(all_images)/255.0#normalization
all_labels=np.array(all_labels)
trainx,testx,trainy,testy=tts(all_images,all_labels,test_size=0.2)
model=KNN(n_neighbors=53).fit(trainx,trainy)
y_predict=model.predict(testx)
print(model.score(testx,testy))
print(confusion_matrix(testy,y_predict))
videoObject=cv2.VideoCapture(1)
letters_and_numbers=list(chars)
while True:
    _,frame=videoObject.read()
    cv2.rectangle(frame,(100,100),(400,400),(0,255,0))
    cv2.imshow('VIDeo',frame)
    roi=frame[100:400,100:400]
    roimg=cv2.resize(roi,(128,128)).flatten()
    new_roimg=np.array(roimg)/255.0
    value=model.predict([new_roimg])
    print(letters_and_numbers[value[0]])
    cv2.putText(frame,letters_and_numbers[value[0]],(50,50),cv2.FONT_HERSHEY_DUPLEX,5,(0,255,0),10)
    if cv2.waitKey(1) & 0xFF==ord('q'):
        break
```

```python
videoObject.release()
cv2.destroyAllWindows()
```
