```python
import matplotlib.pyplot as plt
friends=['a','b','c','d','e']
cscores=[97,87,96,76,93]
sscores=[98,78,96,94,95]
width=0.35
index=[0,1,2,3,4]
fig=plt.subplots(figsize=(10,7))
p1=plt.bar(index,cscores,width)
p2=plt.bar(index,sscores,width,bottom=cscores)
plt.xticks(index,friends)
plt.legend((p1[0],p2[0]),('Coding','Science'))
plt.show()
```

```python
import matplotlib.pyplot as plt
days=['1/1/1948','1/2/1948','1/3/1948','1/4/1948','1/5/1948']
first=[10,6,7,7,7]
second=[8,4,4,4,3]
third=[7,3,2,2,0]
bwidth=0.25
index1=[0,1,2,3,4]
index2=[]
for x in index1:
  index2.append(x+bwidth)
index3=[x+bwidth for x in index2]
fig=plt.subplots(figsize=(20,7))
p1=plt.bar(index1,first,width=bwidth,edgecolor='black',label='Max')
p2=plt.bar(index2,second,width=bwidth,edgecolor='black',label='Mean')
p3=plt.bar(index3,third,width=bwidth,edgecolor='black',label='Mid')
plt.xticks(index1,days)
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import random
x=[]
y=[]
z=[]
for n in range(100):
    x.append(random.randint(1,100))
    y.append(random.randint(1,100))
    z.append(random.randint(1,100))
ax=plt.axes(projection ="3d")
ax.scatter(x,y,z,color='red',marker='o')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
ax.set_title('100 random points')
plt.show()
```

```python
import matplotlib.pyplot as plt
from datetime import datetime, timedelta
import random
start_date=datetime(2019,8,21)
days=7
contest_days=[]
for i in range(days):
    contest_days.append(start_date+timedelta(days=i))
p1=[random.randint(5,15) for a in range(days)]
p2=[random.randint(4,18) for b in range(days)]
p3=[random.randint(6,14) for c in range(days)]
plt.figure(figsize=(12,6))
plt.grid(True)
plt.plot_date(contest_days,p1,marker='o',label='participant 1',color='blue')
plt.plot_date(contest_days,p2,marker='x',label='participant 2',color='red')
plt.plot_date(contest_days,p3,marker='s',label='participant 3',color='green')
plt.gcf().autofmt_xdate()
plt.xlabel('Day of Contest')
plt.ylabel('Number of Pies Eaten')
plt.title('pies eaten over time')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.dates as mpl_dates
from datetime import datetime, timedelta
import random
start_date = datetime(2019, 8, 21)
days=7
contest_days=[]
for i in range(days):
    contest_days.append(start_date + timedelta(days=i))
p1 = [random.randint(5, 15) for a in range(days)]
p2 = [random.randint(4, 18) for b in range(days)]
p3 = [random.randint(6, 14) for c in range(days)]
plt.figure(figsize=(12, 6))
plt.grid(True)
print('Enter a name')
name = input()
print('Pies eaten:')
p = int(input())
dateformat = mpl_dates.DateFormatter('%d-%m-%y')
plt.gca().xaxis.set_major_formatter(dateformat)
plt.plot_date([contest_days[0]], [p], marker='*', markersize=10, label=name, color='purple')
plt.plot_date(contest_days, p1, marker='o', linestyle='-', label='participant 1', color='blue')
plt.plot_date(contest_days, p2, marker='x', linestyle='-', label='participant 2', color='red')
plt.plot_date(contest_days, p3, marker='s', linestyle='-', label='participant 3', color='green')
plt.gcf().autofmt_xdate()
plt.xlabel('Day of Contest')
plt.ylabel('Number of Pies Eaten')
plt.title('Pies Eaten Over Time')
plt.legend()
plt.tight_layout()
plt.show()
```
