```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,r2_score
import matplotlib.pyplot as plt
x=np.linspace(1,10,1000).reshape(-1,1)
y=np.linspace(1,10,1000)+np.random.random(1000 )
plt.scatter(x,y,color='blue')
model=LinearRegression()
model.fit(x,y)
y_predict=model.predict(x)
plt.scatter(x,y)
plt.plot(x,y_predict,color='red')
plt.show()
print(model.coef_,model.intercept_)
value=model.predict([[20]])
print(mean_squared_error(y,y_predict))
print(r2_score(y,y_predict))
```

The reason I need to use .frame is that the fetch_california_housing function, even with as_frame=True, returns a special object called a Bunch, which is like a container that holds different parts of the dataset, including the features, the target, and the full DataFrame.
The drop() method is a pandas DataFrame method, not a Bunch method. I have to access the actual DataFrame inside the Bunch container by using .frame to use pandas functions.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.model_selection import train_test_split as tts
from sklearn.datasets import fetch_california_housing
df=fetch_california_housing(as_frame=True).frame
x=df.drop('MedHouseVal',axis=1)
y=df['MedHouseVal']
trainx,testx,trainy,testy=tts(x,y,test_size=0.2,random_state=42)
model=LinearRegression().fit(trainx,trainy)
y_predict=model.predict(testx)
print(model.coef_,model.intercept_)
print(mean_squared_error(testy,y_predict))
print(r2_score(testy,y_predict))
```
