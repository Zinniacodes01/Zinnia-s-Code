```python
import gradio as gr

with gr.Blocks() as demo:
    gr.Markdown("Advanced Gradio App in Colab")

    with gr.Row():
        name_input = gr.Textbox(label="Enter your name")
        age_slider = gr.Slider(minimum=1, maximum=100, label="Select your age")

    output_text = gr.Textbox(label="Result")
    submit_btn = gr.Button("Submit")

    # Define the action logic
    def complex_greet(name, age):
        return f"{name} is {age} years old!"

    submit_btn.click(fn=complex_greet, inputs=[name_input, age_slider], outputs=output_text)

demo.launch(share=True)
```

```python
import gradio as gr

with gr.Blocks() as demo:
    name = gr.Textbox(label="Name")
    output = gr.Textbox(label="Output")
    btn = gr.Button("Submit")

    # Event binding
    btn.click(fn=lambda x: f"Hello {x}", inputs=name, outputs=output)

demo.launch()
```
