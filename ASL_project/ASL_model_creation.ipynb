```python
all_images=[]
all_labels=[]
base_folder_path='/content/drive/MyDrive/ASL_new/American'
chars="0123456789abcdefghijklmnopqrstuvwxyz"
label_mapping={}
index=0
for char in chars:
    label_mapping[char]=index
    index+=1
print("Loading and preprocessing images...")
print(label_mapping)
for folder in label_mapping:
    folder_path=os.path.join(base_folder_path, folder)
    print(folder_path)
    if not os.path.exists(folder_path): continue
    for image_name in os.listdir(folder_path):
        try:
            img = cv2.imread(os.path.join(folder_path, image_name), cv2.IMREAD_GRAYSCALE)
            img = cv2.resize(img,(64,64))
            all_images.append(img.flatten())
            all_labels.append(label_mapping[folder])
        except Exception:
            continue
```

```python
print('And now for the easy part!')
all_images=np.array(all_images)/255.0#normalization
all_labels=np.array(all_labels)
train_ds=tf.keras.utils.image_dataset_from_directory(base_folder_path,validation_split=0.2,subset="training",seed=42,image_size=(64,64),batch_size=32,color_mode='grayscale')
val_ds=tf.keras.utils.image_dataset_from_directory(base_folder_path,validation_split=0.2,subset="validation",seed=42,image_size=(64,64),batch_size=32,color_mode='grayscale')
normalization_layer=tf.keras.layers.Rescaling(1./255)
train_ds=train_ds.map(lambda x,y:(normalization_layer(x),y))
val_ds=val_ds.map(lambda x,y:(normalization_layer(x),y))
model=tf.keras.Sequential([
    tf.keras.layers.Input(shape=(64,64,1)),
    tf.keras.layers.Conv2D(32,(3,3),activation='relu'),
    tf.keras.layers.MaxPooling2D((2,2)),
    tf.keras.layers.Conv2D(64,(3,3),activation='relu'),
    tf.keras.layers.MaxPooling2D((2,2)),
    tf.keras.layers.Conv2D(128,(3,3),activation='relu'),
    tf.keras.layers.MaxPooling2D((2,2)),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(128,activation='relu'),
    tf.keras.layers.Dropout(0.5),
    tf.keras.layers.Dense(36,activation='softmax')
])
model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
history=model.fit(train_ds,validation_data=val_ds,epochs=15)
#train model(75+ accuracy)  ✅
#test model(save image of my hand) ✅
#store model ✅
#live prediction
#try with landmarks
```

```python
loss,accuracy=model.evaluate(val_ds)
print("Validation Loss:",loss)
print("Validation Accuracy:",accuracy*100,"%")
if accuracy>=0.75:
    print("Heck yes! We did it!")
else:
    print("Aw no. Welp, back to the drawing board.")
```

```python
import os
import cv2
import numpy as np
import pandas as pd
import tensorflow as tf
import matplotlib.pyplot as plt
base_folder_path = '/kaggle/input/name-of-your-asl-dataset/American'
train_ds = tf.keras.utils.image_dataset_from_directory(
    base_folder_path,
    validation_split=0.2,
    subset="training",
    seed=42,
    image_size=(64, 64),
    batch_size=32,
    color_mode='grayscale'
)
val_ds = tf.keras.utils.image_dataset_from_directory(
    base_folder_path,
    validation_split=0.2,
    subset="validation",
    seed=42,
    image_size=(64, 64),
    batch_size=32,
    color_mode='grayscale'
)
normalization_layer = tf.keras.layers.Rescaling(1./255)
train_ds = train_ds.map(lambda x, y: (normalization_layer(x), y))
val_ds = val_ds.map(lambda x, y: (normalization_layer(x), y))
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(64, 64, 1)),
    tf.keras.layers.Conv2D(32, (3, 3), activation='relu'),
    tf.keras.layers.MaxPooling2D((2, 2)),
    tf.keras.layers.Conv2D(64, (3, 3), activation='relu'),
    tf.keras.layers.MaxPooling2D((2, 2)),
    tf.keras.layers.Conv2D(128, (3, 3), activation='relu'),
    tf.keras.layers.MaxPooling2D((2, 2)),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dropout(0.5),
    tf.keras.layers.Dense(36, activation='softmax')
])
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
history = model.fit(train_ds, validation_data=val_ds, epochs=15)
loss, accuracy = model.evaluate(val_ds)
rounded_loss = round(loss, 4)
percentage_accuracy = round(accuracy*100, 2)
print("\nFinal Validation Loss:", rounded_loss)
print("Final Validation Accuracy:", percentage_accuracy, "%")
if accuracy >= 0.75:
    print("Heck yes! We did it!")
else:
    print("Aw no. Welp, back to the drawing board.")
model.save('asl_model.keras')
print("Model saved successfully and is available in the Output files section!")
```
