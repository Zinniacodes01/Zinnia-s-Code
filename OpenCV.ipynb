Image Operations

```python
import cv2 as a
import matplotlib.pyplot as plt
img=a.imread("cat.jpg")
print(img.shape)#different shades of 3 colors
img=a.cvtColor(img,a.COLOR_BGR2RGB)
img=a.cvtColor(img,a.COLOR_RGB2GRAY)
plt.imshow(img,cmap="gray")
print(img.shape)#now all in one shade of black
```

```python
import cv2 as a
import matplotlib.pyplot as plt
img=a.imread("cat.jpg")
img=a.cvtColor(img,a.COLOR_BGR2RGB)
img=a.resize(img,(100,100))
plt.imshow(img)
```

```python
import cv2 as a
import numpy as np
import matplotlib.pyplot as plt
img=a.imread("cat.jpg")
kernel=np.ones((6,6))
img=a.erode(img,kernel)
img=a.cvtColor(img,a.COLOR_BGR2RGB)
#img=a.resize(img,(100,100))
plt.imshow(img)
```

```python
import cv2 as a
import numpy as np
import matplotlib.pyplot as plt
img=a.imread("cat.jpg")
kernel=np.ones((10,10))
img=a.dilate(img,kernel)
img=a.cvtColor(img,a.COLOR_BGR2RGB)
#img=a.resize(img,(100,100))
plt.imshow(img)
```

```python
import cv2 as a
import matplotlib.pyplot as plt
img=a.imread("cat.jpg")
img=a.cvtColor(img,a.COLOR_BGR2RGB)
img=a.cvtColor(img,a.COLOR_RGB2GRAY)
_,threshimg=a.threshold(img,105,255,a.THRESH_BINARY)
plt.imshow(threshimg,cmap="gray")
```

```python
import cv2 as a
import matplotlib.pyplot as plt
img=a.imread("cat.jpg")
img=a.cvtColor(img,a.COLOR_BGR2RGB)
img=a.bitwise_not(img)
plt.imshow(img)
```
